## Unit chart classification

```python
def _unit_chart_type_classification(unit_chart_info):
    """
    Given a unit_chart_info, classify the units into different types,
    and return the indices of units by type or combo of types.
    """
    dead_units_indices = []
    max_num_clusters = np.max(unit_chart_info[:, 1])  # global max used for setting xaxis.
    num_clusters = np.zeros(max_num_clusters+1)
    cluster_sizes = []
    cluster_peaks = []
    border_cell_indices = []
    place_cells_indices = []
    direction_cell_indices = []
    active_no_type_indices = []

    for unit_index in range(unit_chart_info.shape[0]):
        if unit_chart_info[unit_index, 0] == 0:
            dead_units_indices.append(unit_index)
        else:
            num_clusters[int(unit_chart_info[unit_index, 1])] += 1
            cluster_sizes.extend(unit_chart_info[unit_index, 2])
            cluster_peaks.extend(unit_chart_info[unit_index, 3])

            if unit_chart_info[unit_index, 1] > 0:
                place_cells_indices.append(unit_index)
                is_place_cell = True
            else:
                is_place_cell = False

            if unit_chart_info[unit_index, 10] > 0.47:
                direction_cell_indices.append(unit_index)
                is_direction_cell = True
            else:
                is_direction_cell = False

            if unit_chart_info[unit_index, 9] > 0.5:
                border_cell_indices.append(unit_index)
                is_border_cell = True
            else:
                is_border_cell = False

            if not (is_place_cell or is_direction_cell or is_border_cell):
                active_no_type_indices.append(unit_index)

    # plot
    n_dead_units = len(dead_units_indices)
    n_active_units = unit_chart_info.shape[0] - n_dead_units

    # Collect the indices of units that are all three types
    # (place + border + direction)
    place_border_direction_cells_indices = \
        list(set(place_cells_indices) & set(border_cell_indices) & set(direction_cell_indices))
    
    # Collect the indices of units that are two types (inc. three types)
    # (place + border cells)
    # (place + direction cells)
    # (border + direction cells)
    place_and_border_cells_indices = \
        list(set(place_cells_indices) & set(border_cell_indices))
    place_and_direction_cells_indices = \
        list(set(place_cells_indices) & set(direction_cell_indices))
    border_and_direction_cells_indices = \
        list(set(border_cell_indices) & set(direction_cell_indices))
    
    # Collect the indices of units that are only two types
    # (place  + border - direction),
    # (place  + direction   - border),
    # (border + direction   - place)
    place_and_border_not_direction_cells_indices = \
        list(set(place_and_border_cells_indices) - set(place_border_direction_cells_indices))
    place_and_direction_not_border_cells_indices = \
        list(set(place_and_direction_cells_indices) - set(place_border_direction_cells_indices))
    border_and_direction_not_place_cells_indices = \
        list(set(border_and_direction_cells_indices) - set(place_border_direction_cells_indices))
    
    # Collect the indices of units that are exclusive 
    # place cells, 
    # border cells, 
    # direction cells
    exclusive_place_cells_indices = \
        list(set(place_cells_indices) - (set(place_and_border_cells_indices) | set(place_and_direction_cells_indices)))
    exclusive_border_cells_indices = \
        list(set(border_cell_indices) - (set(place_and_border_cells_indices) | set(border_and_direction_cells_indices)))
    exclusive_direction_cells_indices = \
        list(set(direction_cell_indices) - (set(place_and_direction_cells_indices) | set(border_and_direction_cells_indices)))

    results =  {
        'dead_units_indices': dead_units_indices,
        'place_border_direction_cells_indices': place_border_direction_cells_indices,
        'place_and_border_not_direction_cells_indices': place_and_border_not_direction_cells_indices,
        'place_and_direction_not_border_cells_indices': place_and_direction_not_border_cells_indices,
        'border_and_direction_not_place_cells_indices': border_and_direction_not_place_cells_indices,
        'exclusive_place_cells_indices': exclusive_place_cells_indices,
        'exclusive_border_cells_indices': exclusive_border_cells_indices,
        'exclusive_direction_cells_indices': exclusive_direction_cells_indices,
        'active_no_type_indices': active_no_type_indices,
    }
    
    assert unit_chart_info.shape[0] == sum([len(v) for v in results.values()])

    # Check all values are mutually exclusive
    for key, value in results.items():
        for key2, value2 in results.items():
            if key != key2:
                assert len(set(value) & set(value2)) == 0, f'{key} and {key2} have common elements'

    return results
```

```python
import numpy as np
from sklearn.metrics import r2_score
from scipy.optimize import minimize
import seaborn as sns
import matplotlib.pyplot as plt
from tqdm import tqdm

# Func_tion for head direction
def f_const(bins):
    return np.ones(bins.shape)

# 2D Gaussian function
def gaussian_2d(x, y, mu_x, mu_y, sigma_x, sigma_y, rho, A):
    # Compute covariance matrix
    cov_matrix = np.array([
        [sigma_x**2, rho * sigma_x * sigma_y],
        [rho * sigma_x * sigma_y, sigma_y**2]
    ])
    cov_matrix += np.eye(2) * 1e-6
    # Check for singular covariance matrix
    try:
        inv_cov = np.linalg.inv(cov_matrix)
        det_cov = np.linalg.det(cov_matrix)
        if det_cov <= 0:  # Ensure determinant is positive
            raise ValueError("Covariance matrix is singular or non-positive definite.")
    except np.linalg.LinAlgError:
        inv_cov = np.eye(2)  # Default to identity matrix
        det_cov = 1.0

    x_grid, y_grid = np.meshgrid(x, y, indexing='ij') 
    diff = np.stack([x_grid - mu_x, y_grid - mu_y], axis=-1)
    
    exp_term = np.einsum('...i,ij,...j->...', diff, inv_cov, diff)
    exp_term = np.clip(exp_term, -100, 100)  # Clip to prevent overflow
    gaussian = A * np.exp(-0.5 * exp_term) / (2 * np.pi * np.sqrt(det_cov))
    #gaussian = A * np.exp(-0.5 * exp_term)
    #return gaussian
    return np.nan_to_num(gaussian)  # Replace NaNs with 0

# Model for place cell firing rates
def place_cell_model(x, y, hd, mu_x, mu_y, sigma_x, sigma_y, rho, A):
    f_xy = gaussian_2d(x, y, mu_x, mu_y, sigma_x, sigma_y, rho, A)
    f_hd_vals = f_const(hd)
    return f_xy[..., None] * f_hd_vals

def model_error(params, x, y, hd, firing_rate):
    mu_x, mu_y, sigma_x, sigma_y, rho, A = params
    predicted_rate = place_cell_model(x, y, hd, mu_x, mu_y, sigma_x, sigma_y, rho, A)
    error = np.nanmean((firing_rate - predicted_rate)**2)
    return error if np.isfinite(error) else np.inf

n_bins = 17
n_rotations = 17

n_neurons_to_sample = 100

x = np.arange(0, n_bins)
y = np.arange(0, n_bins)
theta = np.arange(0, n_rotations)

r2_per_layer = []
pred_ratemaps_per_layer = []
fitted_params_per_layer = []

for layer_idx, layer in enumerate(layers):
    print(layer)

    place_cells = layer_unit_indices[layer_idx][:n_neurons_to_sample]

    n_neurons = len(place_cells)

    r2_results = np.zeros(n_neurons)
    predicted_ratemaps = np.zeros((n_neurons, n_bins, n_bins, n_rotations))
    fitted_params = np.zeros((n_neurons, 6))

    for n, neuron_idx in enumerate(place_cells):
        heatmap = layer_reps[layer_idx][:, neuron_idx].reshape((n_bins, n_bins, n_rotations))
    
        initial_params = [np.mean(x), np.mean(y), np.std(x), np.std(y), 0, np.max(heatmap)]  # rho starts at 0
        
        try:
            bounds = [
                (0, n_bins),      # mu_x: within the range of x
                (0, n_bins),      # mu_y: within the range of y
                (1e-3, n_bins),   # sigma_x: strictly positive
                (1e-3, n_bins),   # sigma_y: strictly positive
                (-0.5, 0.5),     # rho: within [-1, 1] but avoid exactly ±1,
                (0, None)
            ]
    
            result = minimize(
                model_error,
                initial_params,
                args=(x, y, theta, heatmap),
                bounds=bounds
            )
        except ValueError as e:
            print(f"Neuron {n}: Optimization failed with error {e}")
            r2_results[n] = np.nan
            continue
    
        mu_x_opt, mu_y_opt, sigma_x_opt, sigma_y_opt, rho_opt, A_opt = result.x
        fitted_params[n] = result.x
    
        optimized_rate = place_cell_model(x, y, theta, mu_x_opt, mu_y_opt, sigma_x_opt, sigma_y_opt, rho_opt, A_opt)
        predicted_ratemaps[n] = optimized_rate

        if np.all(predicted_ratemaps[n] == 0.):
            print("all zero")
            print("heatmap fucked?", np.any(heatmap < 0))
        
        if not np.any(np.isnan(optimized_rate)):
            r_squared = r2_score(heatmap.flatten(), optimized_rate.flatten())
            r2_results[n] = r_squared if np.isfinite(r_squared) else np.nan
        else:
            print('nan')
            r2_results[n] = np.nan

    r2_per_layer.append(r2_results)
    pred_ratemaps_per_layer.append(predicted_ratemaps)
    fitted_params_per_layer.append(fitted_params)

    np.save(f'{model_name}_{layer}_r2_gaussian_results.npy', r2_results)
```

```python
import numpy as np
from sklearn.metrics import r2_score
from scipy.optimize import minimize
import seaborn as sns
import matplotlib.pyplot as plt
from tqdm import tqdm

# 2D Gaussian function
def gaussian_2d(x, y, mu_x, mu_y, sigma_x, sigma_y, rho, A):
    cov_matrix = np.array([
        [sigma_x**2, rho * sigma_x * sigma_y],
        [rho * sigma_x * sigma_y, sigma_y**2]
    ])
    cov_matrix += np.eye(2) * 1e-6
    try:
        inv_cov = np.linalg.inv(cov_matrix)
        det_cov = np.linalg.det(cov_matrix)
        if det_cov <= 0:
            raise ValueError("Covariance matrix is singular or non-positive definite.")
    except np.linalg.LinAlgError:
        inv_cov = np.eye(2)
        det_cov = 1.0

    x_grid, y_grid = np.meshgrid(x, y, indexing='ij') 
    diff = np.stack([x_grid - mu_x, y_grid - mu_y], axis=-1)
    exp_term = np.einsum('...i,ij,...j->...', diff, inv_cov, diff)
    exp_term = np.clip(exp_term, -100, 100)
    gaussian = A * np.exp(-0.5 * exp_term) / (2 * np.pi * np.sqrt(det_cov))
    return np.nan_to_num(gaussian)

# Gaussian function for head direction
def gaussian_hd(bins, mu_hd, sigma_hd, A_hd):
    return A_hd * np.exp(-0.5 * ((bins - mu_hd) / sigma_hd) ** 2)

# Model for place cell firing rates
def place_cell_model(x, y, hd, mu_x, mu_y, sigma_x, sigma_y, rho, A_xy, mu_hd, sigma_hd, A_hd):
    f_xy = gaussian_2d(x, y, mu_x, mu_y, sigma_x, sigma_y, rho, A_xy)
    f_hd_vals = gaussian_hd(hd, mu_hd, sigma_hd, A_hd)
    return f_xy[..., None] * f_hd_vals

def model_error(params, x, y, hd, firing_rate):
    mu_x, mu_y, sigma_x, sigma_y, rho, A_xy, mu_hd, sigma_hd, A_hd = params
    predicted_rate = place_cell_model(x, y, hd, mu_x, mu_y, sigma_x, sigma_y, rho, A_xy, mu_hd, sigma_hd, A_hd)
    error = np.nanmean((firing_rate - predicted_rate) ** 2)
    return error if np.isfinite(error) else np.inf

n_bins = 17
n_rotations = 17

n_neurons_to_sample = 100

x = np.arange(0, n_bins)
y = np.arange(0, n_bins)
theta = np.arange(0, n_rotations)

r2_per_layer = []
pred_ratemaps_per_layer = []
fitted_params_per_layer = []

for layer_idx, layer in enumerate(layers):
    print(layer)

    place_cells = layer_unit_indices[layer_idx][:n_neurons_to_sample]

    n_neurons = len(place_cells)

    r2_results = np.zeros(n_neurons)
    predicted_ratemaps = np.zeros((n_neurons, n_bins, n_bins, n_rotations))
    fitted_params = np.zeros((n_neurons, 9))  # Updated to 9 parameters

    for n, neuron_idx in enumerate(place_cells):
        heatmap = layer_reps[layer_idx][:, neuron_idx].reshape((n_bins, n_bins, n_rotations))

        initial_params = [
            np.mean(x),          # mu_x
            np.mean(y),          # mu_y
            np.std(x),           # sigma_x
            np.std(y),           # sigma_y
            0,                   # rho
            np.max(heatmap),     # A_xy
            np.mean(theta),      # mu_hd
            np.std(theta),       # sigma_hd
            1.0                  # A_hd
        ]

        try:
            bounds = [
                (0, n_bins),              # mu_x
                (0, n_bins),              # mu_y
                (1e-3, n_bins),           # sigma_x
                (1e-3, n_bins),           # sigma_y
                (-0.5, 0.5),              # rho
                (0, None),                # A_xy
                (0, n_rotations),         # mu_hd
                (1e-3, n_rotations),      # sigma_hd
                (0, None)                 # A_hd
            ]

            result = minimize(
                model_error,
                initial_params,
                args=(x, y, theta, heatmap),
                bounds=bounds
            )
        except ValueError as e:
            print(f"Neuron {n}: Optimization failed with error {e}")
            r2_results[n] = np.nan
            continue

        mu_x_opt, mu_y_opt, sigma_x_opt, sigma_y_opt, rho_opt, A_xy_opt, mu_hd_opt, sigma_hd_opt, A_hd_opt = result.x
        fitted_params[n] = result.x

        optimized_rate = place_cell_model(x, y, theta, mu_x_opt, mu_y_opt, sigma_x_opt, sigma_y_opt, rho_opt, A_xy_opt, mu_hd_opt, sigma_hd_opt, A_hd_opt)
        predicted_ratemaps[n] = optimized_rate

        if np.all(predicted_ratemaps[n] == 0.):
            print("All zero predicted rates.")
            print("Any negative values in heatmap?", np.any(heatmap < 0))

        if not np.any(np.isnan(optimized_rate)):
            r_squared = r2_score(heatmap.flatten(), optimized_rate.flatten())
            r2_results[n] = r_squared if np.isfinite(r_squared) else np.nan
        else:
            print('NaN encountered in optimized rate.')
            r2_results[n] = np.nan

    r2_per_layer.append(r2_results)
    pred_ratemaps_per_layer.append(predicted_ratemaps)
    fitted_params_per_layer.append(fitted_params)

    np.save(f'{model_name}_{layer}_r2_gaussian_results.npy', r2_results)
```

## Quantifying variance

```python
def quantify_variance_1d(heatmap_test, heatmap_train, verbose=False):
    # heatmap_test and heatmap_train are average spikes/sec per behavioral bin
    
    test_mean = np.nanmean(heatmap_test)
    train_mean = np.nanmean(heatmap_train)

    n_bins = heatmap_test.shape[0]

    inter_var_test = np.nansum((heatmap_test - test_mean)**2)
    inter_var_train = np.nansum((heatmap_train - train_mean)**2)

    intra_var_all = np.nansum((heatmap_test - heatmap_train)**2)
    intra_var_means = n_bins * (test_mean - train_mean)**2

    if verbose:
        print(f"sum((test - test_mean)^2)={inter_var_test:.4f}")
        print(f"sum((train - train_mean)^2)={inter_var_train:.4f}")
        print(f"sum((test - train)^2)={intra_var_all:.4f}")
        print(f"n_bins * (test_mean - train_mean)^2={intra_var_means:.4f}")
        print()
    
    estimator = 0.5 * (
        inter_var_test +
        inter_var_train -
        intra_var_all +
        intra_var_means
    )
    
    return estimator, inter_var_test, inter_var_train, intra_var_all, intra_var_means


def quantify_variance(heatmap_test, heatmap_train, axis=0, verbose=False):
    """
    Quantify the variance due to a certain variable in a cross-validated manner.
    heatmaps are 2D, and are average spikes/sec across visits to each behavioral bin.
    Negative values indicate dominance of inherent variance over dimension-explained variance.
    """

    test_mean = np.nanmean(heatmap_test, axis=axis)
    train_mean = np.nanmean(heatmap_train, axis=axis)

    # Expand along `axis` so we can subtract
    if axis == 0:
        test_mean_expanded = test_mean[np.newaxis, :]
        train_mean_expanded = train_mean[np.newaxis, :]
    else:
        test_mean_expanded = test_mean[:, np.newaxis]
        train_mean_expanded = train_mean[:, np.newaxis]

    n_bins = heatmap_test.shape[axis]

    # Now compute the variance terms with occupancy weighting
    inter_var_test = np.nansum((heatmap_test - test_mean_expanded)**2)
    inter_var_train = np.nansum((heatmap_train - train_mean_expanded)**2)

    intra_var_all = np.nansum((heatmap_test - heatmap_train)**2)
    intra_var_means = n_bins * np.nansum((test_mean - train_mean)**2)

    if verbose:
        print(f"axis={axis} sum((test - test_mean)^2)={inter_var_test:.4f}")
        print(f"axis={axis} sum((train - train_mean)^2)={inter_var_train:.4f}")
        print(f"axis={axis} sum((test - train)^2)={intra_var_all:.4f}")
        print(f"axis={axis} n_bins * sum((test_mean - train_mean)^2)={intra_var_means:.4f}")
        print()
    
    estimator = 0.5 * (
        inter_var_test +
        inter_var_train -
        intra_var_all +
        intra_var_means
    )
    
    return estimator, inter_var_test, inter_var_train, intra_var_all, intra_var_means
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.io import loadmat
from scipy.io.matlab._mio5_params import mat_struct

# Define a function to extract MATLAB structs into Python dictionaries and arrays
def extract_matlab_obj(obj):
    if not (isinstance(obj, mat_struct) or isinstance(obj, dict)):
        return obj

    out = {}
    obj = obj.__dict__ if isinstance(obj, mat_struct) else obj

    for key, val in obj.items():
        if key in ['_fieldnames', 'type', 'time_unit', 'data_unit', 'name']:
            continue

        if isinstance(val, mat_struct) or isinstance(val, dict):
            out[key] = extract_matlab_obj(val)
        elif isinstance(val, np.ndarray):
            arr = [extract_matlab_obj(v) for v in val]
            out[key] = np.array(arr)
        else:
            out[key] = val

    return out


dataset = 'dusz/Dataset_6'
session = 'A1806-190307'

# Load data
epochs = loadmat(current_file_path / f"data/{dataset}/{session}/Data/BehavEpochs.mat", 
                 struct_as_record=False, squeeze_me=True)
epoch_data = extract_matlab_obj(epochs['wakeEp'])

tracking_data = pd.read_csv(current_file_path / f"data/{dataset}/{session}/Data/Tracking_data.csv")
tracking_data.columns = ['timestamp', 'x', 'y', 'hd', 'pitch', 'roll']

positions = tracking_data[['timestamp', 'x', 'y', 'hd']].to_numpy()
position_mask = (positions[:, 0] > epoch_data['start']) & (positions[:, 0] < epoch_data['stop'])
positions = positions[position_mask]

# Normalize positions
positions[:, 3] = (positions[:, 3] + 360) % 360
positions[:, 1] += abs(np.min(positions[:, 1]))
positions[:, 2] += abs(np.min(positions[:, 2]))

def get_md_from_xy(x, y):
    dx = x[1:] - x[:-1]
    dy = y[1:] - y[:-1]
    direction = (np.degrees(np.arctan2(dy, dx)) + 360) % 360
    direction = np.concatenate([direction, [direction[-1]]])
    return direction

def rolling_mean(data, window_size=5):
    kernel = np.ones(window_size) / window_size
    return np.convolve(data, kernel, mode='same')

# Smooth x and y positions
x = rolling_mean(positions[:, 1], 75)
y = rolling_mean(positions[:, 2], 75)

# Compute movement direction
movement_dir = get_md_from_xy(x, y)

time_window = 7000

# Create subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 10), gridspec_kw={'height_ratios': [1, 1]})

# Head direction plot (top left)
sc = axes[0, 0].scatter(positions[:time_window, 0], positions[:time_window, 3], 
                        c=positions[:time_window, 0], cmap='viridis')
cbar = fig.colorbar(sc, ax=axes[0, 0])
cbar.set_label("Time index")
axes[0, 0].set_xlabel("Time")
axes[0, 0].set_ylabel("HD (degrees)")
axes[0, 0].set_title("Head direction")

# Movement direction plot (top right)
sc = axes[0, 1].scatter(positions[:time_window, 0], movement_dir[:time_window], 
                        c=positions[:time_window, 0], cmap='viridis')
cbar = fig.colorbar(sc, ax=axes[0, 1])
cbar.set_label("Time index")
axes[0, 1].set_xlabel("Time")
axes[0, 1].set_ylabel("MD (degrees)")
axes[0, 1].set_title("Movement direction")

# Trajectory plot (bottom, spanning both columns)
ax_traj = fig.add_subplot(2, 1, 2)
sc = ax_traj.scatter(positions[:time_window, 1], positions[:time_window, 2], 
                     c=positions[:time_window, 0], cmap='viridis')
cbar = fig.colorbar(sc, ax=ax_traj)
cbar.set_label("Time index")
ax_traj.set_xlabel("X Position")
ax_traj.set_ylabel("Y Position")
ax_traj.set_title("Trajectory")

# Adjust layout
plt.tight_layout()
plt.show()
```

```python
# the same as above but for MD/HD tuning

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
import seaborn as sns
from tqdm import tqdm
import plotly.express as px
import plotly.graph_objects as go
import os

# ------------------
# Create a Plotly scatter figure with text= index so the index appears next to each point
# ------------------
def create_scatter_plot(df, zoom=False, zoom_range=2000):
    title = f'MD vs HD, dataset session, {len(df)} neurons'
    title += ' (zoomed)' if zoom else ''

    # Create a column for text that only shows the index if the point is red (non_hd == 1)
    df = df.copy()  # to avoid modifying the original DataFrame
    df['red_index'] = [str(idx) if non_hd == 1 else '' for idx, non_hd in zip(df.index, df['non_hd'])]

    # Use the new 'red_index' column for text and hover name
    fig = px.scatter(
        df,
        x='md',
        y='hd',
        title=title,
        labels={'md': 'V (MD)', 'hd': 'V (HD)'},
        text='red_index',
        hover_name='red_index',
        color=df['non_hd'].astype(str),
        color_discrete_map={'0': 'blue', '1': 'red'}
    )

    max_val = max(df['md'].max(), df['hd'].max())
    if zoom:
        max_val = max(max_val, zoom_range)
    fig.add_shape(
        type='line',
        x0=0, y0=0,
        x1=max_val, y1=max_val,
        line=dict(color='red', dash='dash')
    )

    fig.update_traces(textposition='top center')
    if zoom:
        fig.update_layout(
            width=800,
            height=800,
            xaxis=dict(title="V (MD)", range=[0, zoom_range]),
            yaxis=dict(title="V (HD)", range=[0, zoom_range])
        )
    else:
        fig.update_layout(
            width=800,
            height=800,
            xaxis_title="V (MD)",
            yaxis_title="V (HD)"
        )
    return fig


# ------------------
# Plotly 3D volume for each neuron
# ------------------
def create_3d_plot(data_3d, neuron_idx):
    # data_3d shape: (n_bins, n_bins, n_rotations)
    x, y, z = np.meshgrid(
        np.arange(data_3d.shape[0]),
        np.arange(data_3d.shape[1]),
        np.arange(data_3d.shape[2])
    )

    fig_3d = go.Figure(data=go.Volume(
        x=x.flatten(),
        y=y.flatten(),
        z=z.flatten(),
        value=data_3d.flatten(),
        opacity=0.3,
        isomin=0,
        isomax=np.max(data_3d),
        surface_count=20,
        colorscale='Viridis'
    ))

    # Tilt the camera at an angle
    fig_3d.update_layout(
        title=f"Neuron {neuron_idx}",
        scene=dict(
            xaxis_title='X',
            yaxis_title='Y',
            zaxis_title='Rotation',
            camera=dict(
                eye=dict(x=1.25, y=1.25, z=1.25)  # tweak as desired
            )
        ),
        width=600,
        height=600
    )
    return fig_3d

# ------------------
# Polar plot for a single neuron
# ------------------
def create_polar_plot(neuron_idx, mean_firing_rates, which="HD"):
    # mean_firing_rates: shape (n_hd_bins, n_neurons)
    fig, ax = plt.subplots(subplot_kw=dict(projection='polar'), figsize=(4,4))
    
    num_bins = mean_firing_rates.shape[0]
    angles = np.linspace(0, 2*np.pi, num_bins, endpoint=False)
    frates = mean_firing_rates[:, neuron_idx]

    ax.plot(angles, frates, linewidth=2)
    ax.fill(angles, frates, alpha=0.25)
    ax.set_theta_zero_location('N')
    ax.set_theta_direction(-1)
    ax.set_title(f"{which} Polar Tuning - Neuron {neuron_idx}", fontsize=10)

    plt.tight_layout()
    return fig

dataset = 'dusz'
dataset_index = 6

data_path = current_file_path / f'results/{dataset}/Dataset_{dataset_index}'

final_results_dict = {'md': [], 'hd': [], 'non_hd': []}
total_firing_rates = []

for session in tqdm(os.listdir(data_path)):
    if 'A' not in session:
        continue

    session_path = data_path / session

    # Load non-head-direction neuron indexes (1: non-HD, 0: HD)
    non_hd = np.load(session_path / 'non_hd_neuron_indexes.npy')

    hd_rate_map_test = np.load(session_path / 'hd_md_rate_map_fold0_17_17.npy')
    hd_rate_map_train = np.load(session_path / 'hd_md_rate_map_fold1_17_17.npy')
    
    n_md_bins, n_hd_bins, n_neurons = hd_rate_map_test.shape
    
    # Compute final results for each neuron in this session
    for idx in range(n_neurons):
        heatmap_test_flat = hd_rate_map_test[:, :, idx]
        heatmap_train_flat = hd_rate_map_train[:, :, idx]
    
        var_md = quantify_variance(heatmap_test_flat, heatmap_train_flat, axis=0)
        var_hd = quantify_variance(heatmap_test_flat, heatmap_train_flat, axis=1)
    
        final_results_dict['md'].append(var_md[0])
        final_results_dict['hd'].append(var_hd[0])
        final_results_dict['non_hd'].append(non_hd[idx])
    
    # Load firing rates for later use (if needed)
    firing_rates = np.load(session_path / 'hd_md_rate_map_17_17.npy')
    total_firing_rates.append(firing_rates)

total_firing_rates = np.concatenate(total_firing_rates, axis=-1)
n_neurons = total_firing_rates.shape[2]
mean_md = np.nanmean(total_firing_rates, axis=0)
mean_hd = np.nanmean(total_firing_rates, axis=1)

final_results_dict['md'] = np.array(final_results_dict['md'])
final_results_dict['hd'] = np.array(final_results_dict['hd'])
final_results_dict['non_hd'] = np.array(final_results_dict['non_hd'])

final_results = pd.DataFrame(final_results_dict)

# ------------------
# Generate the PDF
# ------------------
output_path = current_file_path / f"results/{dataset}/Dataset_{dataset_index}"

Path(output_path).mkdir(parents=True, exist_ok=True)

output_pdf = output_path / "md_hd_results_summary.pdf"

scatter_fig = create_scatter_plot(final_results)
zoomed_fig = create_scatter_plot(final_results, zoom=True)

from matplotlib.backends.backend_pdf import PdfPages
import matplotlib.image as mpimg

# Save both scatter plots to the PDF
with PdfPages(output_pdf) as pdf:
    # --- First page with ratio text + original scatter plot ---
    fig_first_page = plt.figure(figsize=(8, 11))
    
    # Save Plotly scatter to a temporary image, then load into the matplotlib figure
    scatter_fig.write_image("temp_scatter.png")
    scatter_img = plt.imread("temp_scatter.png")
    plt.axes([0.05, 0.05, 0.9, 0.6])  # [left, bottom, width, height]
    plt.imshow(scatter_img)
    plt.axis('off')
    
    pdf.savefig(fig_first_page)
    plt.close(fig_first_page)

    # --- Second page with zoomed scatter plot ---
    zoomed_fig.write_image("temp_scatter_zoomed.png")
    fig_second_page = plt.figure(figsize=(8, 11))
    zoomed_img = plt.imread("temp_scatter_zoomed.png")
    plt.imshow(zoomed_img)
    plt.axis('off')
    pdf.savefig(fig_second_page)
    plt.close(fig_second_page)

    # -- One page per neuron: polar plot + 3D volume --
    for neuron_idx in tqdm(range(n_neurons)):
        # First polar plot
        fig_polar_1 = create_polar_plot(neuron_idx, mean_md, which='HD')
        fig_polar_1.savefig("temp_polar1.png", dpi=150)
        plt.close(fig_polar_1)
        polar_img_1 = mpimg.imread("temp_polar1.png")
    
        # Second polar plot
        fig_polar_2 = create_polar_plot(neuron_idx, mean_hd, which='MD')
        fig_polar_2.savefig("temp_polar2.png", dpi=150)
        plt.close(fig_polar_2)
        polar_img_2 = mpimg.imread("temp_polar2.png")
    
        # Combine both polar plots into a single figure
        fig_combined = plt.figure(figsize=(8, 4))
    
        ax1 = fig_combined.add_subplot(1, 2, 1)
        ax1.imshow(polar_img_1)
        ax1.axis('off')
    
        ax2 = fig_combined.add_subplot(1, 2, 2)
        ax2.imshow(polar_img_2)
        ax2.axis('off')
    
        plt.suptitle(f"Neuron {neuron_idx}", fontsize=14)
        pdf.savefig(fig_combined)
        plt.close(fig_combined)

print(f"PDF saved to {output_pdf}")
```

```python
import os
import sys
import numpy as np

def get_md_from_xy(positions, smoothing_window=75):
    def rolling_mean(positions):
        kernel = np.ones(smoothing_window) / smoothing_window
        return np.convolve(positions, kernel, mode='same')
        
    x = rolling_mean(positions[:, 1])
    y = rolling_mean(positions[:, 2])
    
    dx = x[1:] - x[:-1]
    dy = y[1:] - y[:-1]

    direction = np.degrees(np.arctan2(dy, dx))
    direction = (direction + 360) % 360

    # Make movement direction have the same length as the original data
    direction = np.concatenate([direction, [direction[-1]]])

    return direction

def circ_mean(angles_deg):
    """
    Compute the circular mean of angles (in degrees).
    """
    angles_rad = np.deg2rad(angles_deg)
    mean_angle_rad = np.arctan2(np.mean(np.sin(angles_rad)), np.mean(np.cos(angles_rad)))
    mean_angle_deg = np.rad2deg(mean_angle_rad) % 360
    return mean_angle_deg

def circular_correlation(angles1, angles2):
    """
    Computes the circular-circular correlation coefficient between two arrays of angles (in degrees)
    using the formula:

        r = sum(sin(theta - theta_bar)*sin(phi - phi_bar)) /
            sqrt(sum(sin^2(theta - theta_bar)) * sum(sin^2(phi - phi_bar)))

    where theta_bar and phi_bar are the circular means.
    """
    mean1 = circ_mean(angles1)
    mean2 = circ_mean(angles2)
    
    # Convert differences to radians so that np.sin() works properly.
    theta_diff = np.deg2rad(angles1 - mean1)
    phi_diff = np.deg2rad(angles2 - mean2)
    
    numerator = np.sum(np.sin(theta_diff) * np.sin(phi_diff))
    denominator = np.sqrt(np.sum(np.sin(theta_diff)**2) * np.sum(np.sin(phi_diff)**2))
    
    if denominator == 0:
        return np.nan  # Avoid division by zero.
    return numerator / denominator

def process_session(session_path):
    """
    Process a single session folder. Looks for a "Data" subfolder and within it
    a Tracking_data.csv file. Loads the data, normalizes x and y to be non-negative,
    computes movement direction from x and y, and returns the circular correlation
    between head direction and movement direction.
    """
    data_folder = os.path.join(session_path, "Data")
    if not os.path.isdir(data_folder):
        print(f"Skipping {os.path.basename(session_path)}: 'Data' folder not found.")
        return None
    
    csv_file = os.path.join(data_folder, "Tracking_data.csv")
    if not os.path.isfile(csv_file):
        print(f"Skipping {os.path.basename(session_path)}: 'Tracking_data.csv' not found in 'Data'.")
        return None
    
    try:
        # Assuming no header and comma-delimited values.
        data = np.loadtxt(csv_file, delimiter=',')
    except Exception as e:
        print(f"Error loading {csv_file}: {e}")
        return None

    # Expecting at least 4 columns: timestamp (0), x (1), y (2), head direction (3)
    if data.shape[1] < 4:
        print(f"Skipping {os.path.basename(session_path)}: Data has fewer than 4 columns.")
        return None
    
    # Normalize x and y so they are non-negative.
    x = data[:, 1]
    y = data[:, 2]
    data[:, 1] = x + abs(np.min(x))
    data[:, 2] = y + abs(np.min(y))
    
    # Extract head direction from column 3.
    head_direction = data[:, 3]
    
    # Compute movement direction using the provided function.
    movement_direction = get_md_from_xy(data)
    
    if len(head_direction) != len(movement_direction):
        print(f"Skipping {os.path.basename(session_path)}: Mismatch in data lengths.")
        return None
    
    # Compute the circular-circular correlation.
    corr = circular_correlation(head_direction, movement_direction)
    return corr

def main(dataset_folder):
    # List all immediate subfolders in the dataset folder; each is a session.
    session_dirs = [os.path.join(dataset_folder, d) for d in os.listdir(dataset_folder)
                    if os.path.isdir(os.path.join(dataset_folder, d))]
    
    correlations = []
    for session in session_dirs:
        corr = process_session(session)
        if corr is not None:
            print(f"Session: {os.path.basename(session)}, Circular Correlation: {corr}")
            correlations.append(corr)
    
    if correlations:
        avg_corr = np.nanmean(correlations)
        print(f"\nAverage Circular Correlation across sessions: {avg_corr}")
    else:
        print("No valid sessions processed.")

if __name__ == "__main__":
    if len(sys.argv) < 2:
        print("Usage: python calc_circular_corr.py /path/to/dataset_folder")
        sys.exit(1)
    dataset_folder = sys.argv[1]
    main(dataset_folder)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
import seaborn as sns
from tqdm import tqdm
import os
from pathlib import Path

# ---------------------------------------------------------
# Same variance-quantification function you already have
# ---------------------------------------------------------
def quantify_variance(heatmap_test, heatmap_train, axis):
    '''
    Cross-validated variance estimator.
    heatmap_test, heatmap_train: 2D arrays with the same shape
    axis: dimension along which you want to measure variance
          (0 = HD dimension, 1 = MD dimension, for example)
    '''
    test_mean = np.nanmean(heatmap_test, axis=axis)
    if axis == 0:
        test_mean_expanded = np.repeat(test_mean[np.newaxis, :], heatmap_test.shape[0], axis=0)
    else:
        test_mean_expanded = np.repeat(test_mean[:, np.newaxis], heatmap_test.shape[1], axis=1)

    train_mean = np.nanmean(heatmap_train, axis=axis)
    if axis == 0:
        train_mean_expanded = np.repeat(train_mean[np.newaxis, :], heatmap_train.shape[0], axis=0)
    else:
        train_mean_expanded = np.repeat(train_mean[:, np.newaxis], heatmap_train.shape[1], axis=1)

    n_bins = heatmap_test.shape[axis]
    test_total_mean = np.nanmean(heatmap_test)
    train_total_mean = np.nanmean(heatmap_train)

    estimator = 0.5 * (
        np.nansum((heatmap_test - test_mean_expanded)**2)
      + np.nansum((heatmap_train - train_mean_expanded)**2)
      - np.nansum((heatmap_test - heatmap_train)**2)
      + n_bins * ((test_total_mean - train_total_mean)**2)
    )

    return estimator

# ---------------------------------------------------------
# Simple scatter plot of var(HD) vs. var(MD)
# ---------------------------------------------------------
def create_scatter_plot_md_hd(df, dataset='', session='', zoom=False, zoom_range=1000):
    """
    Plots a scatter of var(MD) vs. var(HD).
    df must have columns ['hd', 'md'].
    """
    title_str = f"HD vs MD Var, {dataset} {session} ({len(df)} neurons)"
    if zoom:
        title_str += " (zoomed)"

    fig, ax = plt.subplots(figsize=(6,6))
    ax.scatter(df['md'], df['hd'], alpha=0.7, c='blue', edgecolor='k')

    # Optionally draw a diagonal if you want to compare them
    max_val = max(df['md'].max(), df['hd'].max())
    if zoom:
        max_val = max(zoom_range, max_val)
    ax.plot([0, max_val], [0, max_val], 'r--')

    ax.set_xlabel('Variance (MD)')
    ax.set_ylabel('Variance (HD)')
    ax.set_title(title_str)

    if zoom:
        ax.set_xlim([0, zoom_range])
        ax.set_ylim([0, zoom_range])

    fig.tight_layout()
    return fig

# ---------------------------------------------------------
# 2D heatmap of HD vs MD for a single neuron
# ---------------------------------------------------------
def create_hd_md_heatmap(test_map, train_map, neuron_idx=0):
    """
    Creates a matplotlib figure with 2 side-by-side heatmaps:
    Left: fold0 (test_map), Right: fold1 (train_map).
    Both are 2D arrays shaped (n_hd_bins, n_md_bins).
    """
    fig, axes = plt.subplots(1, 2, figsize=(10, 4))

    sns.heatmap(test_map, ax=axes[0], cmap='viridis')
    axes[0].set_title(f"Fold0: Neuron {neuron_idx}")
    axes[0].set_xlabel("MD bin")
    axes[0].set_ylabel("HD bin")

    sns.heatmap(train_map, ax=axes[1], cmap='viridis')
    axes[1].set_title(f"Fold1: Neuron {neuron_idx}")
    axes[1].set_xlabel("MD bin")
    axes[1].set_ylabel("HD bin")

    fig.tight_layout()
    return fig

# ---------------------------------------------------------
# Main analysis code
# ---------------------------------------------------------
dataset = 'dusz'
dataset_index = 6

current_file_path = Path().resolve()
data_path = current_file_path / f'results/{dataset}/Dataset_{dataset_index}'

# We'll store cross-validated variance results here
final_results_dict = {'hd': [], 'md': []}

# We'll also collect the raw rate maps to plot them later
fold0_list = []
fold1_list = []

for session in tqdm(os.listdir(data_path)):
    if 'A' not in session:
        # Skip anything not an actual data folder
        continue

    # Load the new 3D arrays: shape (n_hd_bins, n_md_bins, n_neurons)
    hd_md_rate_map_fold0 = np.load(
        data_path / session / "hd_md_rate_map_fold0_17x17.npy"
    )
    hd_md_rate_map_fold1 = np.load(
        data_path / session / "hd_md_rate_map_fold1_17x17.npy"
    )

    fold0_list.append(hd_md_rate_map_fold0)
    fold1_list.append(hd_md_rate_map_fold1)

# Concatenate along axis=-1 if the code is consistent 
# and we want to combine sessions. That means we stack the neuron dimension
# if the sessions truly have the same (n_hd_bins, n_md_bins) shape each time.
# If the number of neurons differs by session, handle carefully.
fold0_all = np.concatenate(fold0_list, axis=-1)  # shape = (n_hd_bins, n_md_bins, total_neurons)
fold1_all = np.concatenate(fold1_list, axis=-1)  # same shape

n_hd_bins, n_md_bins, n_neurons = fold0_all.shape

# Compute variance for each neuron
for idx in range(n_neurons):
    test_map = fold0_all[..., idx]   # shape: (n_hd_bins, n_md_bins)
    train_map = fold1_all[..., idx]  # shape: (n_hd_bins, n_md_bins)

    var_hd = quantify_variance(test_map, train_map, axis=0)[0]  # collapses over MD dimension
    var_md = quantify_variance(test_map, train_map, axis=1)[0]  # collapses over HD dimension

    final_results_dict['hd'].append(var_hd)
    final_results_dict['md'].append(var_md)

# Make a dataframe for easier plotting
final_results = pd.DataFrame(final_results_dict)

# ---------------------------------------
# Save everything to a PDF
# ---------------------------------------
output_path = current_file_path / f"results/{dataset}/Dataset_{dataset_index}"
output_path.mkdir(parents=True, exist_ok=True)
output_pdf = output_path / "hd_md_tuning_summary.pdf"

with PdfPages(output_pdf) as pdf:
    # Page 1: scatter plot of var(HD) vs var(MD)
    fig_scatter = create_scatter_plot_md_hd(final_results, dataset=dataset, session=f"Dataset_{dataset_index}")
    pdf.savefig(fig_scatter)
    plt.close(fig_scatter)

    # Page 2: optional zoomed scatter
    fig_scatter_zoom = create_scatter_plot_md_hd(final_results, dataset=dataset, session=f"Dataset_{dataset_index}", zoom=True, zoom_range=500)
    pdf.savefig(fig_scatter_zoom)
    plt.close(fig_scatter_zoom)

    # Then one page per neuron: 2D heatmap fold0 vs fold1
    for neuron_idx in tqdm(range(n_neurons)):
        test_map = fold0_all[..., neuron_idx]
        train_map = fold1_all[..., neuron_idx]
        fig_map = create_hd_md_heatmap(test_map, train_map, neuron_idx=neuron_idx)
        pdf.savefig(fig_map)
        plt.close(fig_map)

print(f"PDF saved to {output_pdf}")
```

```python
def _compute_single_heatmap_directional_scores(activation_maps):
    """
    Args:
        `activation_maps` correspond to the un-summed activation maps
        for a single unit across rotations \in (n_locations, n_rotations)

    - num of angular bins in Banino here becomes `n_rotations`.
    - based on Banino eq, we need to convert each n_rotations to 
        `alpha_i` which is angle.
    - the intensity `beta_i` of an angle is the average activation 
        across all locations for that angle.
    """
    activation_maps = (activation_maps - np.min(activation_maps)) / (np.max(activation_maps) - np.min(activation_maps))
    
    num_bins = activation_maps.shape[1]
    alphas = np.linspace(0, 2*np.pi, num=n_rotations, endpoint=False)
    betas = np.mean(activation_maps, axis=0) # mean direction for this spatial bin

    polar_plot_coords = [] # (n_rotations, 2)
    per_rotation_vector_length = []
    for alpha_i, beta_i in zip(alphas, betas):
        polar_plot_coords.append(
            [beta_i*np.cos(alpha_i), beta_i*np.sin(alpha_i)]
        )
        per_rotation_vector_length.append(
            np.linalg.norm([beta_i*np.cos(alpha_i), beta_i*np.sin(alpha_i)])
        )
    
    r_normed_by_beta = np.sum(
        np.array(polar_plot_coords), axis=0) / np.sum(betas)

    mean_vector_length = np.linalg.norm(r_normed_by_beta)
    return mean_vector_length, per_rotation_vector_length
    
```

```python
import numpy as np
from sklearn.metrics import r2_score
from scipy.optimize import minimize
import seaborn as sns
import matplotlib.pyplot as plt
from tqdm import tqdm

envLen = 200
n_bins = 17
n_rotations = 16
binWidth = envLen/n_bins

smoothRateMaps = np.load(f'/home/mag/Space/results/pfeiffer/smoothRateMaps.npy') # smoothed xy map: n_bins, n_bins, n_neurons
hdRateMap = np.load(f'/home/mag/Space/results/pfeiffer/hdRateMap.npy') # firing rate per hd: n_bins, n_bins, n_rotations, n_neurons
hdOccupancyMap = np.load(f'/home/mag/Space/results/pfeiffer/hdOccupancyMap.npy') # visit counts: n_bins, n_bins, n_rotations

# Func_tion for head direction
def f_const(bins):
    return np.ones(bins.shape)

# 2D Gaussian function
def gaussian_2d(x, y, mu_x, mu_y, sigma_x, sigma_y, rho, A):
    # Compute covariance matrix
    cov_matrix = np.array([
        [sigma_x**2, 0],
        [0, sigma_y**2]
    ])
    cov_matrix += np.eye(2) * 1e-6
    # Check for singular covariance matrix
    try:
        inv_cov = np.linalg.inv(cov_matrix)
        det_cov = np.linalg.det(cov_matrix)
        if det_cov <= 0:  # Ensure determinant is positive
            raise ValueError("Covariance matrix is singular or non-positive definite.")
    except np.linalg.LinAlgError:
        inv_cov = np.eye(2)  # Default to identity matrix
        det_cov = 1.0

    x_grid, y_grid = np.meshgrid(x, y, indexing='ij') 
    diff = np.stack([x_grid - mu_x, y_grid - mu_y], axis=-1)
    
    exp_term = np.einsum('...i,ij,...j->...', diff, inv_cov, diff)
    exp_term = np.clip(exp_term, -100, 100)  # Clip to prevent overflow
    gaussian = A * np.exp(-0.5 * exp_term) / (2 * np.pi * np.sqrt(det_cov))
    #gaussian = A * np.exp(-0.5 * exp_term)
    #return gaussian
    return np.nan_to_num(gaussian)  # Replace NaNs with 0

# Model for place cell firing rates
def place_cell_model(x, y, hd, mu_x, mu_y, sigma_x, sigma_y, rho, A):
    f_xy = gaussian_2d(x, y, mu_x, mu_y, sigma_x, sigma_y, rho, A)
    f_hd_vals = f_const(hd)
    return f_xy[..., None] * f_hd_vals

def model_error(params, x, y, hd, firing_rate):
    mu_x, mu_y, sigma_x, sigma_y, rho, A = params
    predicted_rate = place_cell_model(x, y, hd, mu_x, mu_y, sigma_x, sigma_y, rho, A)
    error = np.nanmean((firing_rate - predicted_rate)**2)
    return error if np.isfinite(error) else np.inf


n_bins = 17
n_rotations = 16
n_neurons = hdRateMap.shape[2]

x = np.arange(0, n_bins)
y = np.arange(0, n_bins)
theta = np.arange(0, n_rotations)

# Initialize results array
r2_results = np.zeros(n_neurons)
predicted_ratemaps = np.zeros((n_neurons, n_bins, n_bins, n_rotations))
fitted_params = np.zeros((n_neurons, 6))

# Iterate over neurons
for neuron in tqdm(range(n_neurons)):
    # Compute firing rate heatmap
    heatmap = hdRateMap[:, :, neuron, :] / (hdOccupancyMap + 1e-8)  # Avoid division by zero
    heatmap = np.nan_to_num(heatmap)  # Replace NaNs with 0

    # Initial guesses for optimization
    initial_params = [np.mean(x), np.mean(y), np.std(x), np.std(y), 0, np.max(heatmap)]  # rho starts at 0
    
    # Optimize parameters
    try:
        bounds = [
            (0, n_bins),      # mu_x: within the range of x
            (0, n_bins),      # mu_y: within the range of y
            (1e-3, n_bins),   # sigma_x: strictly positive
            (1e-3, n_bins),   # sigma_y: strictly positive
            (-0.5, 0.5),     # rho: within [-1, 1] but avoid exactly ±1,
            (0, None)
        ]

        result = minimize(
            model_error,
            initial_params,
            args=(x, y, theta, heatmap),
            bounds=bounds
        )
    except ValueError as e:
        print(f"Neuron {neuron}: Optimization failed with error {e}")
        r2_results[neuron] = np.nan
        continue

    # Extract optimized parameters
    mu_x_opt, mu_y_opt, sigma_x_opt, sigma_y_opt, rho_opt, A_opt = result.x
    fitted_params[neuron] = result.x

    # Compute optimized firing rate and R²
    optimized_rate = place_cell_model(x, y, theta, mu_x_opt, mu_y_opt, sigma_x_opt, sigma_y_opt, rho_opt, A_opt)
    predicted_ratemaps[neuron] = optimized_rate
    
    if not np.any(np.isnan(optimized_rate)):
        r_squared = r2_score(heatmap.flatten(), optimized_rate.flatten())
        r2_results[neuron] = r_squared if np.isfinite(r_squared) else np.nan
    else:
        r2_results[neuron] = np.nan

```
